🎉 Algoritmo: organizar uma festa
# Algoritmo para organizar uma festa

print("🎉 ORGANIZAÇÃO DA FESTA 🎉")

convidados = int(input("Quantas pessoas serão convidadas? "))
orcamento = float(input("Qual é o orçamento da festa? R$ "))

# Estimativa de comida e bebida
comida_por_pessoa = 30
bebida_por_pessoa = 15

custo_comida = convidados * comida_por_pessoa
custo_bebida = convidados * bebida_por_pessoa

custo_total = custo_comida + custo_bebida

print("\n--- RESUMO DA FESTA ---")
print("Número de convidados:", convidados)
print("Custo estimado com comida: R$", custo_comida)
print("Custo estimado com bebida: R$", custo_bebida)
print("Custo total estimado: R$", custo_total)

if custo_total <= orcamento:
    print("Orçamento suficiente! Festa aprovada! 🥳")
else:
    print("Atenção: o orçamento não é suficiente.")
    print("Será necessário reduzir os gastos.")
🧠 O que esse algoritmo está fazendo?
Pede o número de convidados.
Pede o orçamento disponível.
Calcula o custo da comida.
Calcula o custo das bebidas.
Soma os custos.
Usa if para verificar se o orçamento é suficiente.
Mostra o resultado na tela.
Por exemplo, se você colocar 20 convidados e R$ 1.000, o programa calcula:
Comida: 20 × 30 = R$ 600
Bebida: 20 × 15 = R$ 300
Total: R$ 900
Orçamento: R$ 1.000
Resultado: Orçamento suficiente! 🎉
Esse é um bom exemplo para você treinar variáveis, input(), cálculos e if/else, que são justamente a base para começar a programar em Python.

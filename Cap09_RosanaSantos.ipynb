```python
meta = float(input("Digite o valor da meta: R$ "))
vendido = float(input("Digite o valor vendido: R$ "))

diferenca = vendido - meta

if vendido >= meta:
  print ("Meta atingida")
  print (f"Você ficou R$ {diferenca:.2f} acima da meta.")
else:
  print ("Meta não atingida")
  print (f"Faltaram R$ {abs (diferenca):.2f} para atingir a meta.")

```

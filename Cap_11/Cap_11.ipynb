Exercício 10

```python
n_alunos = int(input('Quantos alunos têm na turma? '))
if n_alunos <= 0:
  print('A quantidade de alunos deve ser maior que zero')
else:
  soma_medias = 0.0
  aprovados = 0
  recuperacao = 0
  reprovados = 0
  maior_media = -1.0
  menor_media = 11.0
  nome_maior = ""
  nome_menor = ""

  print()
  print('===Processamento da Turma===')

  for i in range(n_alunos):
    nome = input(f'Nome do aluno {i+1}: ')
    P1 = float(input('Nota P1(30%): '))
    P2 = float(input('Nota P2(30%): '))
    P3 = float(input('Nota P3(20%): '))
    Projeto = float(input('Nota Projeto(20%): '))

    media = (P1 * 0.30) + (P2 * 0.30) + (P3 * 0.20) + (Projeto * 0.20)
    soma_medias += media

    if media >= 7.0:
      situacao = 'Aprovado'
      aprovados += 1
    elif media >= 5.0:
      situacao = 'Recuperação'
      recuperacao += 1
    else:
      situacao = 'Reprovado'
      reprovados += 1

    if media > maior_media:
      maior_media = media
      nome_maior = nome
    if media < menor_media:
      menor_media = media
      nome_menor = nome

    print(f' {nome}: média {media:.2f} | {situacao}')

print()
media_turma = soma_medias / n_alunos
print('===Resumo da Turma===')
print(f'Média geral: {media_turma:.2f}')
print(f'Aprovados: {aprovados}')
print(f'Em recuperação: {recuperacao}')
print(f'Reprovados: {reprovados}')
print(f'Maior média: {nome_maior} ({maior_media:.2f})')
print(f'Menor média: {nome_menor} ({menor_media:.2f})')
```

Exercício 11


```python
meta = 5000.00
vendedores = 10
total_vendido = 0.0
bateram_meta = 0
for i in range(1, vendedores + 1):
 venda = float(input(f'Vendas do vendedor {i}: R$ '))
 total_vendido += venda
 if venda >= meta:
    bateram_meta += 1

media = total_vendido / vendedores

print()
print('=== Relatório de Vendas ===')
print(f'Total vendido: R$ {total_vendido:.2f}')
print(f'Média de vendas: R$ {media:.2f}')
print(f'Vendedores que bateram a meta: {bateram_meta}')
print(f'Percentual que bateu a meta: '
 f'{bateram_meta / vendedores:.0%}')
```

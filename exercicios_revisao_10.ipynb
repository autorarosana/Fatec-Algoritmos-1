```python
temperatura = float(input ("Digite a temperatura "))

if temperatura < 15:
  classificacao = "Frio"
elif temperatura < 25:
  classificacao = "Agradável"
else:
  classificacao = "Quente"

print (f"Classificação: {classificacao}")



```

```python
valor_compra = float (input ("Digite o valor da compra: R$ "))

if valor_compra >= 200:
  valor_final = valor_compra * 0.90
else:
  valor_final = valor_compra

print (f"Valor final: R$ {valor_final:.2f}")
```

```python
valor_compra = float (input("Digite o valor da compra: R$"))
regiao = input("Região (local, estadual ou nacional): ").lower()

if valor_compra >= 300:
  frete = 0
else:
  if regiao == "local":
    frete = 10
  elif regiao == "estadual":
    frete = 25
  elif regiao == "nacional":
    frete = 50
  else :
    frete = 0
    print ("Região inválida."
            "Frete não calculado")

total_final = valor_compra + frete

print (f"Valor da compra: R$ {valor_compra:.2f}")
print (f"Valor do frete: R$ {frete:.2f}")
print (f"Total final: R$ {total_final:.2f}")

```

Exercício 10

```python
soma = 0
quantidade = 0

valor = float(input('Digite um valor positivo ou -1 para sair: '))

while valor != -1:
  soma += valor
  quantidade += 1
  valor = float(input('Digite um valor positivo ou -1 para sair): '))

if quantidade > 0:
  media = soma / quantidade
  print (f'Soma: {soma:.2f}')
  print (f'Quantidade: {quantidade}')
  print (f'Média: {media:.2f}')
else:
  print ('Nenhum valor válido foi digitado.')

```

Exercício 15


```python
import random
limite_tentativas = 7
jogar_novamente = 'S'

while jogar_novamente.upper() == 'S':
  numero_secreto = random.randint(1, 100)
  tentativas = 0
  acertou = False

  print('\n' + '=' * 42)
  print('Adivinhe o número secreto entre 1 e 100')
  print(f'Você tem {limite_tentativas} tentativas.')
  print('=' * 42)

  while tentativas < limite_tentativas and not acertou:
    restante = limite_tentativas - tentativas
    chute = int(input(f'[{restante} restante(s)] Seu chute: '))
    tentativas += 1

    if chute == numero_secreto:
      acertou = True
    elif chute < numero_secreto:
      print(f'O número é MAIOR que {chute}.')
    else:
      print(f'O número é MENOR que {chute}.')

  if acertou:
    print(f'Parabéns! Você acertou em {tentativas} tentativas.')
  else:
    print(f'Você perdeu! O número secreto era {numero_secreto}.')

  jogar_novamente = input('Deseja jogar novamente? (S/N): ')

print('\nObrigado por jogar!')
```
